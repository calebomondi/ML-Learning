```python
import pandas as pd

print(pd.__version__)
```

```python
# Series - 1D labeled array capable of holding any data type

## LISTS

data = [102.5, 123.6, 234.7, 324.1, 438.8]
series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])
# Whole series
print(f"series:\n {series}")
# accessing data in series using loc and iloc
print(f"loc: {series.loc['c']}")
print(f"iloc: {series.iloc[2]}")
# updating data in series using loc and iloc
series.loc['c'] = 250.0
print(f"updated series:\n {series}")

# Filtering
belowe_200 = series[series < 200]
above_200 = series[series > 200]
print(f"belowe_200:\n {belowe_200}")
print(f"above_200:\n {above_200}")
```

```python
## DICTIONARIES

calories = {'day1': 102.5, 'day2': 123.6, 'day3': 234.7, 'day4': 324.1, 'day5': 438.8}
series = pd.Series(calories)
# Whole series
print(f"series:\n {series}")
# accessing data in series using loc and iloc
print(f"loc: Day 3 - {series.loc['day3']}")
print(f"iloc: Day 3 - {series.iloc[2]}")
# updating data in series using loc and iloc
series.loc['day3'] += 250.0
print(f"updated series:\n {series}")
# Filtering
belowe_200 = series[series < 200]
above_200 = series[series > 200]
print(f"belowe_200:\n {belowe_200}")
print(f"above_200:\n {above_200}")
```

```python
# DataFrame - 2D labeled tabular data structure with rows and columns of potentially different types
# - Similar to a spreadsheet or SQL table, or a dict of Series objects

data = {
    "Name": ["Spongebob", "Patrick", "Squidward"],
    "Age": [20, 21, 22],
}

df = pd.DataFrame(data, index=['Employee1', 'Employee2', 'Employee3'])

# Add Column
df["Job"] = ["Fry Cook", "Unemployed", "Cashier"]
df["Salary"] = [1500, 0, 2000]

# Add Row
new_rows = pd.DataFrame(
    [{"Name": "Sandy", "Age": 24, "Job": "Engineer", "Salary": 32000},
    {"Name": "Mr. Krabs", "Age": 45, "Job": "Business Owner", "Salary": 100000}],
    index=['Employee4', 'Employee5']
)
df = pd.concat([df, new_rows])

print(f"DataFrame:\n {df}")
```

```python
from pathlib import Path

# READ CSV FILE
csv_path = Path("data/pokemon.csv")
if not csv_path.exists():
    csv_path = Path.cwd().parent / "data" / "pokemon.csv"

df = pd.read_csv(csv_path)
print(f"DataFrame from CSV:\n {df}")
```

```python
# SELECTION BY COLUMN

## By Column Name
print(f"Column Name - Name:\n {df['Name']}")

## By Column Name, Height and Weight
print(f"Column Name - Height and Weight:\n {df[['Name','Height', 'Weight']]}")
```

```python
# SELECTION BY ROWS

df = pd.read_csv(csv_path, index_col="Name")

## By Location (loc)

### Access Pikachu row by name
print(f"Selection by Row - Name:\n {df.loc['Pikachu']}")

### Access multiple rows by name
print(f"\nSelection by Row - Names:\n {df.loc[['Pikachu', 'Bulbasaur']]}")

### Access only height and weight of Pikachu
print(f"\nSelection by Row - Name, Height and Weight:\n {df.loc['Pikachu', ['Height', 'Weight']]}")

### Access Height and Weight of Pokemons between Pikachu and Bulbasaur
print(f"\nSelection by Row - Names, Height and Weight:\n {df.loc['Bulbasaur':'Pikachu', ['Height', 'Weight']]}")

## By Index (iloc)

### Access pokemons from index 0 to 10, getting only Height and Weight
print(f"\nSelection by Index - Range:\n {df.iloc[0:10, 0:3]}")
```

```python
# FILTERING

### Tall pokemon
tall_pokemon = df[df['Height'] > 2.0]
print(f"\nTall Pokemon:\n {tall_pokemon}")

### Tall pokemon with weight greater than 100
tall_heavy_pokemon = df[(df['Height'] > 2.0) & (df['Weight'] > 100)]
print(f"\nTall and Heavy Pokemon:\n {tall_heavy_pokemon}")

### Pokemon whose Type 1 is Grass or Type 2 is Poison
grass_poison_pokemon = df[(df['Type1'] == 'Grass') | (df['Type2'] == 'Poison')]
print(f"\nGrass or Poison Pokemon:\n {grass_poison_pokemon}")
```

```python
# AGGREGATION - Reduces a set of values into a single value
# - Used to summarize and analyze data
# - Often used with the groupby() function

df = pd.read_csv(csv_path, index_col="Name")

## Whole DataFrame

# Mean
print(f"\nMean:\n {df.mean(numeric_only=True)}")
# Sum
print(f"\nSum:\n {df.sum(numeric_only=True)}")
# Count
print(f"\nCount:\n {df.count()}")
# Min
print(f"\nMin:\n {df.min(numeric_only=True)}")
# Max
print(f"\nMax:\n {df.max(numeric_only=True)}")

```

```python
## Single Column

# Mean
print(f"\nMean:\n {df['Height'].mean()}")
# Sum
print(f"\nSum:\n {df['Height'].sum()}")
# Count
print(f"\nCount:\n {df['Height'].count()}")
# Min
print(f"\nMin:\n {df['Height'].min()}")
# Max
print(f"\nMax:\n {df['Height'].max()}")
```

```python
## GroupBy - Splits the data into groups based on some criteria

group = df.groupby('Type1')

print(f"\nGroupBy - Type1:\n {group}")

# Sum
print(f"\nGroupBy - Type1 - Sum:\n {group['Height'].sum()}")
# Count of Pokemons by Type1
print(f"\nGroupBy - Type1 - Count:\n {group['Type1'].count()}")
# Mean Height
print(f"\nGroupBy - Type1 - Height:\n {group['Height'].mean()}")
# Mean Weight
print(f"\nGroupBy - Type1 - Weight:\n {group['Weight'].mean()}")
```

```python
# DATA CLEANING - Process of identifying and correcting errors and inconsistencies in data to improve its quality and reliability

df = pd.read_csv(csv_path)

# print(f"\nDataFrame:\n {df.to_string()}")

# Drop Irrelevant Columns
df = df.drop(columns=["Legendary", "No"])
print(df)

```

```python
## Handle Missing Data

### Remove rows with missing data
df = pd.read_csv(csv_path)

df = df.drop(columns=["Legendary", "No"])
df = df.dropna(subset=["Type2"])
print(df)
```

```python
### Fill in rows with missing values

df = pd.read_csv(csv_path)

df = df.drop(columns=["Legendary", "No"])
df = df.fillna({"Type2": "None"})
print(df)
```

```python
### Fix inconsistent values

df["Type1"] = df["Type1"].replace({"Grass": "GRASS", "Fire": "FIRE"})
print(df)
```

```python
### Standardize Text

df["Name"] = df["Name"].str.lower()
print(df)
```

```python
### Fix data types

df = pd.read_csv(csv_path)
df["Legendary"] = df["Legendary"].astype(bool)
print(df)
```
