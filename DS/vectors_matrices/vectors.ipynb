```python
# ARITHMETIC OPERATIONS 

from typing import List

Vector = List[float]

def add(v: Vector, w: Vector) -> Vector:
    """Adds corresponding vectors"""

    assert len(v) == len(w)
    return [v_i + w_i for v_i, w_i in zip(v, w)]

assert add([1, 2, 3], [4, 5, 6]) == [5, 7, 9]

def subtract(v: Vector, w: Vector) -> Vector:
    """Subtracts corresponding vectors"""

    assert len(v) == len(w)
    return [v_i - w_i for v_i, w_i in zip(v, w)]

assert subtract([5, 7, 9], [4, 5, 6]) == [1, 2, 3]
```

```python
# COMPONENTWISE SUM - create a new vector whose 1st element is the sum of all the first elements. whose 2nd is the sum of all the second elemnts, and so on.

def vector_sum(vectors: List[Vector]) -> Vector:
    """Sums all corresponding elements"""

    # Check that vectors is not empty
    assert vectors, "no vectors provided!"

    # Check the vectors are all same size
    num_elements = len(vectors[0])
    assert all(len(v) == num_elements for v in vectors), "different sizes!"

    # the i-th element of the result is the sum of every vector[i]
    return [sum(vector[i] for vector in vectors) for i in range(num_elements)]

assert vector_sum([[1, 2], [3, 4], [5, 6], [7, 8]]) == [16, 20]
```

```python
# MULTIPLY VECTOR BY SCALAR

def scalar_multiply(c: float, v: Vector) -> Vector:
    """Multiplies every element by x"""

    return [c * v_i for v_i in v]

assert scalar_multiply(2, [1, 2, 3]) == [2, 4, 6]
```

```python
# This allows us to compute the componentwise means of a list

def vector_mean(vectors: List[Vector]) -> vector_sum:
    """Computes element wise average"""

    n = len(vectors)
    return scalar_multiply(1/n, vector_sum(vectors))

assert vector_mean([[1,2], [3,4], [5,6]]) == [3, 4]
```

```python
# DOT PRODUCT - of 2 vectors is the sum of their corresponding COMPONENTWISE PRODUCTS

def dot(v: Vector, w: Vector) -> float:
    """Computes v_i * w_i + ... + v_n * w_n"""

    assert len(v) == len(w)
    return sum(v_i * w_i for v_i, w_i in zip(v, w)) 

assert dot([1, 2, 3], [4, 5, 6]) == 32 # 1 * 4 + 2 * 5 + 3 * 6


# - If w has a magnitud of 1, the dot product measures how far the vector v extends in the w direction
# - It's the length of the vector you'd get if you projected v onto w (Another way of saying it, ...)

# Make's it easier to compute vectors's 'sum of squares'

def sum_of_squares(v: Vector) -> float:
    """Returns v_1 * w_1 + ... + v_n * w_n"""

    return dot(v, v)

assert sum_of_squares([1, 2, 3]) == 14 # 1*1 + 2*2 + 3*3

# which can be used to compute its magnitude (oe length)

import math

def magnitude(v:Vector) -> float:
    """Returns the magnitude (or length) of v"""
    return math.sqrt(sum_of_squares(v))

assert magnitude([3, 4]) == 5
```

```python
# Can now compute the distance between 2 vectors

def squared_distance(v: Vector, w: Vector) -> float:
    """Computes (v_i - w_i) ** 2 + ... + (v_n - w_n) ** 2"""
    return  sum_of_squares(subtract(v, w))

def distance(v: Vector, w: Vector) -> float:
    """Computes distance between v and w"""
    return math.sqrt(squared_distance(v, w))

# Clearer to write as 

def distance(v: Vector, w: Vector) -> float:
    return magnitude(subtract(v, w))
```
