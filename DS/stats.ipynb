```python
# STATISTICS

# - refers to the mathematics and techniques with which we understand data

from typing import List
```

```python
# MEAN - The average of a list of numbers

def mean(xs: List[float]) -> float:
    return sum(xs) / len(xs)

assert mean([1, 2, 3]) == 2
```

```python
# MEDIAN - The middle value in a sorted list of numbers

def _median_odd(xs: List[float]) -> float:
    """If len(xs) is odd, then median is the middle element"""
    return sorted(xs)[len(xs) // 2]

def _median_even(xs: List[float]) -> float:
    """If len(xs) is even, it's the average of the middle two elements"""
    sorted_xs = sorted(xs)
    hi_midpoint = len(xs) // 2 #e.g. 4 // 2 = 2, 6 // 2 = 3
    return (sorted_xs[hi_midpoint - 1] + sorted_xs[hi_midpoint]) / 2

def median(v: List[float]) -> float:
    """Compute the median of a list of numbers"""
    return _median_odd(v) if len(v) % 2 == 1 else _median_even(v)

assert median([1, 10, 2, 9, 5]) == 5
assert median([1, 9, 2, 10]) == (2 + 9) / 2
```

```python
# QUANTILE - The pth-percentile value in a list of numbers

def quantile(xs: List[float], p: float) -> float:
    """Return the pth-percentile value in x"""
    p_index = int(p * len(xs))
    return sorted(xs)[p_index]

assert quantile([1, 2, 3, 4, 5], 0.1) == 1
assert quantile([1, 2, 3, 4, 5], 0.25) == 2
assert quantile([1, 2, 3, 4, 5], 0.5) == 3
assert quantile([1, 2, 3, 4, 5], 0.75) == 4
assert quantile([1, 2, 3, 4, 5], 0.9) == 5
```

```python
# MODE - Most frequently occurring value(s) in a list

from collections import Counter

def mode(xs: List[float]) -> List[float]:
    """Returns a list since there might be more than one mode"""

    counts = Counter(xs) # returns a dictionary of {value: count} pairs
    max_count = max(counts.values()) # finds the highest count
    return [x_i for x_i, count in counts.items() if count == max_count] # returns a list of all values that have the highest count

assert mode([1, 2, 3, 3, 4, 4]) == [3, 4]
assert set(mode([1, 3, 3, 4, 4, 5])) == {3, 4}
```

```python
# DISPERSION - refers to measures of how spread out our data is.

def data_range(xs: List[float]) -> float:
    """Returns the difference between the largest and smallest values"""
    return max(xs) - min(xs)

assert data_range([1, 2, 3, 4, 5]) == 4

# Variance 
# - The average squared distance from the mean
# - Measures how a single varibale deviates from its mean

# - How to calculate variance:
# 1. Find the mean of the data
# 2. For each data point, find the distance from the mean and square it
# 3. Find the average of those squared distances


```

```python
# SUM OF SQUARES - v_1^2 + ... + v_n^2
# - Equivalently dot(v, v); see vectors.ipynb

def sum_of_squares(xs: List[float]) -> float:
    """Returns the sum of the squares of the elements in xs"""
    return sum(x ** 2 for x in xs)

assert sum_of_squares([1, 2, 3]) == 14

def de_mean(xs: List[float]) -> List[float]:
    """Returns the deviations of each element from the mean"""
    x_bar = mean(xs)
    return [x - x_bar for x in xs]

def variance(xs: List[float]) -> float:
    """Returns the variance of a list of numbers"""

    assert len(xs) >= 2, "variance requires at least two elements"

    n = len(xs)
    deviations = de_mean(xs)
    return sum_of_squares(deviations) / (n - 1)

data = [1, 2, 3, 4, 5]
assert variance(data) == 2.5
```

```python
# STANDARD DEVIATION - The square root of the variance

import math

def standard_deviation(xs: List[float]) -> float:
    """Returns the standard deviation of a list of numbers"""
    return math.sqrt(variance(xs))

assert standard_deviation(data) == math.sqrt(variance(data))
```

```python
# Interquartile Range (IQR) - The difference between the 75th and 25th percentiles

def interquartile_range(xs: List[float]) -> float:
    """Returns the interquartile range of a list of numbers"""
    return quantile(xs, 0.75) - quantile(xs, 0.25)

assert interquartile_range([1, 2, 3, 4, 5]) == 2
```

```python
# Correlation and Covariance - all about how two things move together. If one thing goes up, does the other go up or down? How much do they move together?
# - Positive correlation means that as one variable increases, the other variable tends to increase as well. Negative correlation means that as one variable increases, the other variable tends to decrease. Covariance is a measure of how much two random variables vary together.
```

```python
# Covariance - A measure of how two variables vary in tandem from their  means

def covariance(xs: List[float], ys: List[float]) -> float:
    """Returns the covariance between two lists of numbers"""

    assert len(xs) == len(ys), "xs and ys must have the same length"

    xs_deviations = de_mean(xs)
    ys_deviations = de_mean(ys)

    return sum(x * y for x, y in zip(xs_deviations, ys_deviations)) / (len(xs) - 1)

def correlation(xs: List[float], ys: List[float]) -> float:
    """Measures how much xs and ys vary in tandem about their means"""

    stdev_xs = standard_deviation(xs)
    stdev_ys = standard_deviation(ys)

    if stdev_xs > 0 and stdev_ys > 0:
        return covariance(xs, ys) / (stdev_xs * stdev_ys) 
    else:
        return 0 # if no variation, correlation is zero

assert correlation([1, 2, 3], [4, 5, 6]) == 1
assert correlation([1, 2, 3], [6, 5, 4]) == -1
```
